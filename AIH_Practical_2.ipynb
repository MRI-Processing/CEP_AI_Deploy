```python
import gradio as gr
import pickle
import numpy as np

# 1. Load the trained model
model_filename = 'heart_failure_model.pkl'
with open(model_filename, 'rb') as file:
    model = pickle.load(file)

# 2. Define the prediction function
def predict_heart_disease(age, cholesterol):
    # Reshape input to match 2D array expected by scikit-learn
    input_data = np.array([[age, cholesterol]])
    prediction = model.predict(input_data)[0]
    probability = model.predict_proba(input_data)[0][1]

    result = "Heart Disease Detected " if prediction == 1 else "No Heart Disease Detected"
    return f"{result} (Confidence: {probability * 100:.2f}%)"

# 3. Create and launch the Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("# Heart Failure Prediction Assistant")
    gr.Markdown("Input age and cholesterol levels to predict heart disease likelihood using your trained Logistic Regression model.")

    with gr.Row():
        age_input = gr.Number(label="Age (Years)", value=45, precision=0)
        chol_input = gr.Number(label="Cholesterol (mg/dl)", value=200, precision=0)

    output_text = gr.Textbox(label="Prediction Outcome")

    predict_btn = gr.Button("Predict")
    predict_btn.click(fn=predict_heart_disease, inputs=[age_input, chol_input], outputs=output_text)

demo.launch(inline=True)
```
